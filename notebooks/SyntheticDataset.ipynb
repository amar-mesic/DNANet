```python
from DNAnet.data.data_models import Panel
from DNAnet.data.parsing.file_categorization_strategy import SyntheticFileCategorizer
from DNAnet.data.validation.sample_validation_strategy import NFIValidationStrategy
from DNAnet.data.kit_compatibility.lane_standards import InternalSizeStandard
from DNAnet.data.data_models.synthetic_dataset import SyntheticDataset
```

# Create Synthetic Dataset

```python
gf_panel_path = "resources/data/synthetic/SGPanel_Globalfiler_Panel.xml"
data_files_path = "resources/data/synthetic"
reference_genotype_path = "resources/data/synthetic/fixed_ratios_base_template_500_5000_no_preprocess/reference_genotypes"
epg_to_genotypes_mapping_path = "resources/data/synthetic/fixed_ratios_base_template_500_5000_no_preprocess/alleles_to_genotypes_mapping.csv"
file_categorizer = SyntheticFileCategorizer()
sample_validator = NFIValidationStrategy()
```

```python
all_images = SyntheticDataset(
    root_path=data_files_path,
    panel_path=gf_panel_path,
    reference_genotype_path=reference_genotype_path,
    epg_to_genotypes_mapping_path=epg_to_genotypes_mapping_path,
    size_standard=InternalSizeStandard.SYNTHETIC_GENESCAN_600_LIZ.value,
    file_categorization_strategy=file_categorizer,
    sample_validation_strategy=sample_validator,
)
```

```python
len(all_images.sample_files), len(all_images)
```

```python
from config_io import load_model

images = all_images[:30]

unet_model = load_model("config/models/unet.yaml")
unet_model.load("resources/model/current_best_unet/")
predictions = unet_model.predict_batch(images)
```

```python
from DNAnet.evaluation.segmentation.allele_metrics import allele_f1_score, allele_precision, allele_recall

# current issue: we are also retrieving ladder files, and other files not conforming to the standard
print('allele u-net f1: ', allele_f1_score(images, predictions))
print('allele u-net precision: ', allele_precision(images, predictions))
print('allele u-net recall: ', allele_recall(images, predictions))
```
