# Convert HID Files to Numpy Arrays

This notebook loads raw HID files from the specified dataset, extracts the 6 dye lanes per file, and saves each as a numpy array of shape (6, 9461, 1) to the output directory.

```python
import os
import numpy as np
from DNAnet.data.parsing.parse_raw_hid import get_peak_data
from DNAnet.data.parsing.file_categorization_strategy import ProvedItFileCategorizer

# Input and output directories
input_dir = "/Users/amarmesic/Documents/tudelft/thesis/datasets/USE THIS - PROVEDIt_2-5-Person Profiles_3500 5sec_GF29cycles"
output_dir = "/Users/amarmesic/Documents/tudelft/thesis/dna/real_epgs/PROVEDIt_2-5-Person Profiles_3500 5sec_GF29cycles"
os.makedirs(output_dir, exist_ok=True)
```

# Create HID Dataset

```python
# Find all HID files in the input directory
hid_files = []
for root, dirs, files in os.walk(input_dir):
    for file in files:
        if file.lower().endswith('.hid'):
            hid_files.append(os.path.join(root, file))
print(f"Found {len(hid_files)} HID files.")
```

```python
# Only keep sample files using the ProvedItFileCategorizer
categorizer = ProvedItFileCategorizer()
sample_hid_files = [f for f in hid_files if categorizer(f.split(os.sep)[-1]) == "sample"]
print(f"Found {len(sample_hid_files)} sample HID files.")

for hid_path in sample_hid_files:
    arr = get_peak_data(hid_path)
    
    if arr is not None:
        arr = arr[..., np.newaxis]
        arr = arr[:, 4000:9000]
        
        # Use file name (without extension) for output
        out_name = os.path.splitext(os.path.basename(hid_path))[0] + ".npy"
        out_path = os.path.join(output_dir, out_name)
        np.save(out_path, arr)
        print(f"Saved: {out_path} - shape: {arr.shape}")
    else:
        print(f"Failed to process: {hid_path}")
```
