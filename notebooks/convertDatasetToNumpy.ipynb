# Convert ProvedIt and Synthetic Datasets to Numpy Arrays

This notebook loads a ProvedIt dataset and a synthetic dataset, extracts the raw data from each image, and saves each as a `.npy` file in the specified output directory.

```python
import os
import numpy as np
from DNAnet.data.data_models.custom_hid_dataset import CustomHIDDataset
from DNAnet.data.data_models.synthetic_image import SyntheticImage
from DNAnet.data.data_models.dna_models import Panel
from DNAnet.data.kit_compatibility.lane_standards import InternalSizeStandard
from DNAnet.data.parsing.file_categorization_strategy import ProvedItFileCategorizer
from DNAnet.data.validation.sample_validation_strategy import NFIValidationStrategy
from DNAnet.data.data_models.synthetic_dataset import SyntheticDataset
from DNAnet.data.parsing.file_categorization_strategy import SyntheticFileCategorizer

thesis_dir = "/Users/amarmesic/Documents/tudelft/thesis/"

# Input and output directories
provedit_input_dir = f"{thesis_dir}/datasets/USE THIS - PROVEDIt_2-5-Person Profiles_3500 5sec_GF29cycles"
provedit_output_dir = f"{thesis_dir}/dna/preprocessed_epgs/real/PROVEDIt_2p-5p_5sec_GF29cyc_temp_scaled"
os.makedirs(provedit_output_dir, exist_ok=True)

gf_panel_path = "resources/data/ProvedIt/SGPanel_Globalfiler_Panel.xml"
```

## Convert ProvedIt Dataset

```python
provedit_dataset = CustomHIDDataset(
    root_path=provedit_input_dir,
    panel_path=gf_panel_path,
    size_standard=InternalSizeStandard.GENESCAN_600_LIZ.value,
    file_categorization_strategy=ProvedItFileCategorizer(),
    sample_validation_strategy=NFIValidationStrategy(),
)

print(f"Number of ProvedIt images: {len(provedit_dataset)}")
```

```python
for img in provedit_dataset:
    # Extract the raw data as a numpy array
    data = img.data
    # Save to output directory with a unique name
    filename = os.path.splitext(os.path.basename(str(img.path)))[0] + ".npy"
    out_path = os.path.join(provedit_output_dir, filename)
    np.save(out_path, data)
```

## Convert Synthetic Dataset

```python
synthetic_input_dir = f"{thesis_dir}/dna/generated_epgs/fixed_ratios_all_loci_thresh_15"
synthetic_output_dir = f"{thesis_dir}/dna/preprocessed_epgs/synth/fixed_ratios_all_loci_thresh_15"
os.makedirs(synthetic_output_dir, exist_ok=True)
```

```python
synthetic_panel_path = "resources/data/ProvedIt/SGPanel_Globalfiler_Panel.xml"
synthetic_reference_genotype_path = f"{synthetic_input_dir}/reference_genotypes"
synthetic_epg_to_genotypes_mapping_path = f"{synthetic_input_dir}/alleles_to_genotypes_mapping.csv"

synth_dataset = SyntheticDataset(
    root_path=synthetic_input_dir,
    panel_path=gf_panel_path,
    limit=512,
    reference_genotype_path=synthetic_reference_genotype_path,
    epg_to_genotypes_mapping_path=synthetic_epg_to_genotypes_mapping_path,
    size_standard=InternalSizeStandard.SYNTHETIC_GENESCAN_600_LIZ.value,
    file_categorization_strategy=SyntheticFileCategorizer(),
    sample_validation_strategy=NFIValidationStrategy(),
)
```

```python
for img in synth_dataset:
    # Extract the raw data as a numpy array
    data = img.data
    # Save to output directory with a unique name
    filename = os.path.splitext(os.path.basename(str(img.path)))[0] + ".npy"
    out_path = os.path.join(synthetic_output_dir, filename)
    np.save(out_path, data)
```

Both datasets have now been converted to numpy arrays and saved in their respective output directories.
